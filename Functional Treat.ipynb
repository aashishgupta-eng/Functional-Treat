```python
# ============================================
# Data Analyzer and Transformer Program
# ============================================

data = []


def input_data():
    global data
    user = input("\nEnter numbers separated by space: ")
    data = list(map(int, user.split()))
    print("Data Stored Successfully!")


def display_summary():
    if len(data) == 0:
        print("No Data Available!")
        return

    print("\n----- Data Summary -----")
    print("Data      :", data)
    print("Total     :", len(data))
    print("Minimum   :", min(data))
    print("Maximum   :", max(data))
    print("Sum       :", sum(data))
    print("Average   :", round(sum(data) / len(data), 2))


def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)



def calculate_factorial():
    num = int(input("\nEnter Number: "))

    if num < 0:
        print("Factorial not possible for negative numbers.")
    else:
        print(f"Factorial of {num} is =", factorial(num))


def filter_data():
    if len(data) == 0:
        print("No Data Available!")
        return

    value = int(input("Enter Threshold Value: "))

    result = list(filter(lambda x: x >= value, data))

    print("Filtered Data :", result)


def sort_data():
    if len(data) == 0:
        print("No Data Available!")
        return

    print("\n1. Ascending")
    print("2. Descending")

    choice = input("Enter Choice: ")

    if choice == "1":
        print("Sorted Data :", sorted(data))

    elif choice == "2":
        print("Sorted Data :", sorted(data, reverse=True))

    else:
        print("Invalid Choice!")


def statistics():

    minimum = min(data)
    maximum = max(data)
    total = sum(data)
    average = total / len(data)

    return minimum, maximum, total, average


def display_statistics():

    if len(data) == 0:
        print("No Data Available!")
        return

    minimum, maximum, total, average = statistics()

    print("\n----- Statistics -----")
    print("Minimum :", minimum)
    print("Maximum :", maximum)
    print("Total   :", total)
    print("Average :", round(average, 2))


def main():

    while True:

        print("\n==============================")
        print(" Data Analyzer and Transformer Program ")
        print("==============================")

        print("1. Input Data")
        print("2. Display Data Summary")
        print("3. Calculate Factorial")
        print("4. Filter Data by Threshold")
        print("5. Sort Data")
        print("6. Display Dataset Statistics")
        print("7. Exit Program")

        option = input("Enter Your Choice: ")

        if option == "1":
            input_data()

        elif option == "2":
            display_summary()

        elif option == "3":
            calculate_factorial()

        elif option == "4":
            filter_data()

        elif option == "5":
            sort_data()

        elif option == "6":
            display_statistics()

        elif option == "7":
            print("\nThank You! for using the data Analyzer and Transformer Program. Goodbye!✌")
            break

        else:
            print("Invalid Choice! Please Try Again.")



main()
```
